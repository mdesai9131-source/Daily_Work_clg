```python
first_name = "Mahesh"
last_name = "Desai"
```

```python
full_name = first_name + " " + last_name
print("Concat Strint (Join) : ", full_name)
```

```python
print("Ascending Character: ")
print("First Character First name: ", first_name[0])
print("Last Character Full_name: ", full_name[-1])
```

```python
# Slicing 
print(full_name[:6])
print(full_name[7:])
```

```python
# searching Text
print("Searching Mahesh: ", "Mahesh" in full_name)
print("Finding Desai: ", full_name.find("Desai")) # Return Index to start 
```

```python
# Replacing Text
new_name = full_name.replace("Desai", "Savdhariya ")
print("After Replacing: ", new_name)
```

```python
# Changing Case 
print("Upper Case: ", new_name.upper())
print("Lower Case: ", new_name.lower())
print("Title Case: ", new_name.title())
```

```python
# Removing Extra spaces
print("Original: ", new_name)
print("After Removing: ", new_name.strip()) # Agal Pachhal Space Remove
print(new_name.rstrip()) # right side Re
```
